# SP 991

```python
mask_sp = (data['spmode']=='991')

cut = -1
mask_cut = (dcuts[cut]['event'])

x = data[mask_sp & mask_cut]['BpostFitMes']
y = data[mask_sp & mask_cut]['BpostFitMes']

plt.hist(ak.flatten(x), bins=50, range=(5.2,5.3))

;
```

```python
all_hists= bat.create_empty_histograms(hist_defs)

bkg_modes =['991']#, '998']
sig_modes = ['-999']

spmodes = bkg_modes + sig_modes

subset = ['Lambda0_unc_Mass', 'BpostFitMes', 'BpostFitDeltaE']

weights = {'991':1, '998':1, '-999':0.1}

df_cutflow_sp = bat.fill_histograms_v3(data, subset=subset, empty_hists=all_hists, spmodes=spmodes, weights=weights, cuts=dcuts);
```

```python
all_hists['Lambda0_unc_Mass']
```

```python

for cut_name in ['1', '2', '3', '4', '6', '-1']:

    #cut_name = '-1'
    
    bat.plot_histograms(all_hists, vars=subset, bkg_spmodes=bkg_modes, sig_spmodes=sig_modes, \
                        datamodes=None, cut=cut_name, overlay_data=False, only_stacked=True, \
                        save= True, fixed_grid=(1,3), extra_tag='_991_studies');
```
